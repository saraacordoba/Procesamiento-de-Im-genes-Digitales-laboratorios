# TEMA 2: Transformaciones de intensidad. 

```python
import cv2 as cv 
import matplotlib.pyplot as plt 
import numpy as np
```

## I. Histograma: 

La función de opencv para obtener el histograma de una imagen es:

    cv.calcHist(images, channels, mask, histSize, ranges)
    
    - images: lista de imágenes (todas deben ser del mismo tipo y tamaño). Lo habitual es una única imagen.
    - channels: lista de canales usados para calcular el histograma.
    - mask: máscara que define la región sobre la que deseamos calcular el histograma, es opcional.
    - histSize: tamaño de los histogramas en cada dimensión.
    - ranges: rangos de valores

```python
img = cv.imread('gato_poco_contraste.jpg',0)
plt.title('Imagen escala de grises')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)
#plt.imshow(img, cmap = 'gray')
```

Calculamos el histograma usando el canal 0 (al ser una imagen cargada en escala de grises, tienen un solo canal) y tomando 256 valores en el rango [0,255].

```python
hist = cv.calcHist([img], [0], None, [256], [0,255]) #cv.calcHist([imagen], [canal], máscara sobre la imagen, tamaño del histograma (número de bins), rango de valores excluyendo el último)
plt.plot(hist)
```

Se puede observar en el histograma que todos los niveles de gris de la imagen están concentrados en un intervalo muy pequeño y en la parte derecha, luego poco contraste y clara.

Veamos ahora un ejemplo de una imagen a color:

```python
img = cv.imread('playa.jpg') 
plt.title('Imagen a color')
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
```

```python
hist_B = cv.calcHist([img], [0], None, [256], [0,256]) 
hist_G = cv.calcHist([img], [1], None, [256], [0,256]) 
hist_R = cv.calcHist([img], [2], None, [256], [0,256])
plt.plot(hist_R, color = 'red')
plt.plot(hist_G, color = 'green')
plt.plot(hist_B, color = 'blue')
```

## II. Transformaciones lineales:

Se tratan de transformaciones T(r)=a*r+b

```python
img = cv.imread('mujer.jpg',0)
plt.title('Imagen escala de grises con alto contraste')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)
#plt.imshow(img, cmap = 'gray')
```

```python
m = np.min(img)
M = np.max(img)

print('El mínimo valor de intensidad es ',m)
print('El máximo valor de intensidad es ',M)
```

```python
hist = cv.calcHist([img], [0], None, [256], [0,255]) #cv.calcHist([imagen], [canal], máscara sobre la imagen, tamaño del histograma (número de bins), rango de valores excluyendo el último)
plt.plot(hist)
```

Vamos a transformarla para que los niveles de gris ahora esten entre 50 y 255

```python
#Nuevos limites
L1 = 50
L2 = 255

a = (L2 - L1) / (M - m)
b = L1 - a * m

img_new = a * img + b

print(img_new.dtype)

img_new = np.uint8(img_new)#.astype(np.uint8)

print(img_new.dtype)
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) #parámetros que se pueden ajustar

plt.subplot(1,2,1)
plt.title('Imagen original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,2,2)
plt.title('Imagen escala de grises nueva escala')
plt.imshow(img_new, cmap = 'gray', vmin = 0, vmax = 255)
```

```python
hist_1 = cv.calcHist([img], [0], None, [256], [0,256]) 
hist_2 = cv.calcHist([img_new], [0], None, [256], [0,256]) 
plt.plot(hist_1, color = 'black')
plt.plot(hist_2, color = 'grey')
```

## III. Transformación log:

Dada una imagen en escala de grises, la transformación intensidad se define como: 
 
         s = T(r) = C*log(1 + r)
         
siendo C una constante cuyo valor se ajusta en función de los rangos de niveles de grises de la imagen de entrada y salida. 

```python
img = cv.imread('mujer.jpg',0)

M = np.amax(img)
print('El máximo valor de intensidad es ',M)
```

Definimos C de forma que lleve el intervalo [0,217] en el intervalo [0,255], es decir, T(M) = 255.

    T(M) = C*log(1 + M) = 255 -> C = 255/log(1+M)

```python
C = 255/np.log(1 + M.astype('float'))
img_log = C*np.log(1 + img.astype('float')) 
img_log = np.uint8(img_log)
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) #parámetros que se pueden ajustar

plt.subplot(1,2,1)
plt.title('Imagen original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,2,2)
plt.title('Log')
plt.imshow(img_log, cmap = 'gray', vmin = 0, vmax = 255)
```

## IV. Transformación exponencial:

Dada una imagen en escala de grises, la transformación intensidad se define como: 
 
         s = T(r) = e^(r/c)-1
         
siendo C una constante cuyo valor se ajusta en función de los rangos de niveles de grises de la imagen de entrada y salida. 

```python
img = cv.imread('Img_Bajo_Contraste.jpg',0)

M = np.amax(img)
print('El máximo valor de intensidad es ',M)
```

Definimos C de forma que lleve el intervalo [0,217] en el intervalo [0,255], es decir, T(M) = 255.

    T(M) = e^(M/c)-1 = 255 -> c = M/log(256)

```python
c = M.astype('float')/np.log(256)
img_exp = (np.exp(img.astype('float') / c) - 1)
img_exp = np.uint8(img_exp)
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) #parámetros que se pueden ajustar

plt.subplot(1,2,1)
plt.title('Imagen original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,2,2)
plt.title('Exponencial')
plt.imshow(img_exp, cmap = 'gray', vmin = 0, vmax = 255)
```

## V. Corrección gamma:

La transformación viene dada por:  
 
         s = T(r) = C*r^gamma
         
donde la constante C se debe ajustar igual que para el caso anterior. 


```python
img = cv.imread('gato_poco_contraste.jpg',0)

M = np.amax(img)
print('El valor máximo de intensidad es ',M)
```

Definimos C de forma que lleve el intervalo [0,M] en el intervalo [0,255], es decir, T(M) = 255.

    T(M) = C*M^gamma = 255 -> C = 255/M^gamma

Probamos para valores de gamma > 1.

```python
gamma = 2
C = 255/M.astype('float')**gamma
img_gamma2 = C*np.power(img.astype('float'),gamma)
img_gamma2 = np.uint8(img_gamma2)
```

```python
gamma = 5
C = 255/M.astype('float')**gamma
img_gamma5 = C*np.power(img.astype('float'),gamma)
img_gamma5 = np.uint8(img_gamma5)
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) #parámetros que se pueden ajustar

plt.subplot(1,3,1)
plt.title('Imagen original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,3,2)
plt.title('Gamma 2')
plt.imshow(img_gamma2, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,3,3)
plt.title('Gamma 5')
plt.imshow(img_gamma5, cmap = 'gray', vmin = 0, vmax = 255)
```

Probamos ahora para gamma < 1 considerando la imagen de la mujer a la que le aplicamos anteriormente la transformación log:

```python
img = cv.imread('mujer.jpg',0)
M = np.amax(img)
```

```python
gamma = 1/2
C = 255/M.astype('float')**gamma
img_gamma2 = C*np.power(img.astype('float'),gamma)
img_gamma2 = np.uint8(img_gamma2)

gamma = 1/5
C = 255/M.astype('float')**gamma
img_gamma5 = C*np.power(img.astype('float'),gamma)
img_gamma5 = np.uint8(img_gamma5)
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) #parámetros que se pueden ajustar

plt.subplot(1,3,1)
plt.title('Imagen original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,3,2)
plt.title('Gamma 1/2')
plt.imshow(img_gamma2, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,3,3)
plt.title('Gamma 1/5')
plt.imshow(img_gamma5, cmap = 'gray', vmin = 0, vmax = 255)
```

## VI. Ecualización del histograma:

La función de opencv para calcular la imagen ecualizada es:

    cv.equalizeHist(image)
    
La imagen que devuelve es de tipo uint8 y, por tanto, no es necesario transformarla.

```python
img = cv.imread('gato_poco_contraste.jpg',0)
```

```python
img_eq = cv.equalizeHist(img)
```

```python
#plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) 

plt.subplot(1,2,1)
plt.title('Original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,2,2)
plt.title('Ecualización')
plt.imshow(img_eq, cmap = 'gray', vmin = 0, vmax = 255)
```

Calculamos el histograma de la imagen ecualizada y comparamos dicho histograma con el de la imagen expandida del apartado anterior:

```python
hist_inicial = cv.calcHist([img], [0], None, [256], [0,255]) 
hist_eq = cv.calcHist([img_eq], [0], None, [256], [0,255]) 
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 2, top = None, wspace = 0.5, hspace = 0.5) 

plt.subplot(1,2,1)
plt.title('Original')
plt.plot(hist_inicial)


plt.subplot(1,2,2)
plt.title('Ecualización')
plt.plot(hist_eq)
```

## VII. Expansión del histograma:

Se trata de una transformación lineal que lleva, mediante una función lineal (una recta), el intervalo "real" de niveles de gris de la imagen inicial en todo el rango posible de valores, considerando imágenes de 8 bits. 

    T : [m,M] -> [0,255] 
    
siendo m y M los valores máximos y mínimos de intensidad de la imagen. Por tanto T(r) se define como la recta que pasa por los puntos (m,0) y (M,255).

    T(r) = 255/(M-m)*(r-m)

Calculamos los valores mínimo y máximo:

```python
img = cv.imread('gato_poco_contraste.jpg',0)

m = np.min(img)
M = np.max(img)

print('El mínimo valor de intensidad es ',m)
print('El máximo valor de intensidad es ',M)
```

Para aplicar la transformación a la imagen, tenemos que tratar los datos como float y no de tipo uint8. Una vez aplicada la transformación, volvemos a pasar a una matriz tipo uint8:

```python
img_exp = 255/(M-m)*(img.astype('float') - m)  
img_exp = np.uint8(img_exp) 
cv.imwrite('gato_exp.png',img_exp)
```

```python
plt.subplots_adjust(left = None, bottom = None, right = 1.5, top = None, wspace = 0.5, hspace = 0.5) 

plt.subplot(1,2,1)
plt.title('Imagen original')
plt.imshow(img, cmap = 'gray', vmin = 0, vmax = 255)

plt.subplot(1,2,2)
plt.title('Expansión')
plt.imshow(img_exp, cmap = 'gray', vmin = 0, vmax = 255)
```

Representamos los histogramas para comprobar que, efectivamente, la tranbsformación realizada ha expandido los niveles de gris a todo el intervalo [0,255]:

```python
hist_inicial = cv.calcHist([img], [0], None, [256], [0,255]) 
hist_exp = cv.calcHist([img_exp], [0], None, [256], [0,255]) 
```

```python
plt.plot(hist_inicial, color = 'red')
plt.plot(hist_exp, color = 'blue')
```

## VI. Transformación de intensidad a una imagen a color: 

### Ecualización usando RGB:

```python
img = cv.imread('poco_contraste_color.jpg') 
plt.title('Imagen a color')
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB)) 

B,G,R = cv.split(img) #Separamos los canales
```

Realizamos la ecualización a cada canal por separado:

```python
img_B = cv.equalizeHist(B)
img_G = cv.equalizeHist(G)
img_R = cv.equalizeHist(R)

img_eq_RGB = cv.merge([img_R, img_G, img_B])

plt.title('Ecualización usando BGR')
plt.imshow(img_eq_RGB) 
```

### Ecualización usando HSV:

```python
imgHSV = cv.cvtColor(img, cv.COLOR_BGR2HSV) #Paso de BGR a HSV

H,S,V = cv.split(img) #Separamos los canales

plt.title('Canal V')
plt.imshow(V, cmap = 'gray', vmin = 0, vmax = 255) 
```

Realizamos la ecualización al canal V:

```python
V_eq = cv.equalizeHist(V) 

plt.title('Canal V ecualizado')
plt.imshow(V_eq, cmap = 'gray', vmin = 0, vmax = 255) 
```

```python
img_eq_HSV = cv.merge([H,S,V_eq])
img_eq_RGB = cv.cvtColor(img_eq_HSV, cv.COLOR_HSV2RGB) #Pasamos de HSV a RGB 

plt.title('Ecualización usando HSV')
plt.imshow(img_eq_RGB) 
```

Otra forma es pasar de HSV a BGR y luego, al mostrar la imagen, pasar de BGR a RGB:

```python
img_eq_BGR = cv.cvtColor(img_eq_HSV, cv.COLOR_HSV2BGR) #Pasamos de HSV a BGR 

plt.title('Ecualización usando HSV')
plt.imshow(cv.cvtColor(img_eq_BGR, cv.COLOR_BGR2RGB))
```

### Transformación log usando RGB y HSV:

```python
img = cv.imread('noche.jpg') 
plt.title('Imagen a color')
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB)) 
```

En el caso del modelo RGB, esta transformación sí la podemos hacer directamente en la imagen porque se trata de realizar operaciones matriciales. La ecualización no porque está definida expresamente para imágenes, es decir, trabaja con el histograma de la imagen.

```python
M = np.amax(img)
C = 255/np.log(1 + M.astype('float'))
img_log = C*np.log(1 + img.astype('float')) 
img_log = np.uint8(img_log)
```

```python
plt.title('Imagen a color transformada usando BGR')
plt.imshow(cv.cvtColor(img_log, cv.COLOR_BGR2RGB)) 
```

```python
imgHSV = cv.cvtColor(img, cv.COLOR_BGR2HSV) 

H, S, V = cv.split(imgHSV)

C = 255/np.log(1 + np.amax(V).astype('float'))
V_log = C*np.log(1 + V.astype('float'))
V_log = np.uint8(V_log)

img_log_HSV = cv.merge([H,S,V_log])

plt.title('Imagen a color transformada usando HSV')
plt.imshow(cv.cvtColor(img_log_HSV, cv.COLOR_HSV2RGB)) 
```
