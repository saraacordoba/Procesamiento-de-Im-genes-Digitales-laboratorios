## Ejercicio 1 — ¿Qué funciones tiene OpenCV para transformaciones y para obtener matrices?

1.- Busca las funciones que tiene OpenCV para transformaciones y para obtener matrices.

2.- Carga la imagen 'Radiografia.jpg' en escala de grises y realiza una rotación en ella con centro el centro de la imagen.

3.- Realiza la misma transformación, pero esta vez rectifícala de forma que no aparezcan bordes negros en la imagen.



#### Aplicar transformaciones:

    cv.warpAffine: (2×3, afines)
    cv.warpPerspective (3×3, proyectivas)
    cv.remap (mapa denso XY)
    cv.resize.

#### Construir/estimar matrices:
Afin:

    cv.getRotationMatrix2D
    cv.getAffineTransform
    cv.invertAffineTransform
    cv.estimateAffine2D/Partial2D.

Proyectiva: 

    cv.getPerspectiveTransform
    cv.findHomography (RANSAC, LMEDS) a partir de correspondencias.

Otros útiles: cv.getRectSubPix, cv.initUndistortRectifyMap (calibración/rectificación).

```python
img_color = cv.imread('Radiografia.jpg')
img_bw = cv.imread('Radiografia.jpg', 0)

plt.figure(figsize=(10,10))
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(img_color, cv.COLOR_BGR2RGB))

plt.subplot(1,2,2)
plt.imshow(img_bw, cmap='gray')

rows, cols = img_bw.shape
```

```python
#M = cv.getRotationMatrix2D((width/2, height/2), 5, 1)
#cv.warpAffine(img_color, M, (width, height))

matriz_rotacion = cv.getRotationMatrix2D((int(cols//2),int(rows//2)), -5, 1)
Ir = cv.warpAffine(img_bw,matriz_rotacion,(cols,rows))
plt.imshow(Ir, cmap = 'gray')
```

interpolation=

cv.INTER_NEAREST (vecino más próximo)

cv.INTER_LINEAR (bilineal) ⭑ por defecto

cv.INTER_CUBIC (bicúbica, 4×4, más suave)

cv.INTER_AREA (ideal para reducir, promedio por área)

cv.INTER_LANCZOS4 (filtro sinc ventana 8×8, muy alta calidad al ampliar)

(cv.INTER_LINEAR_EXACT en algunas builds)

```python
# Tablero (8x8)
checker = np.zeros((8,8), dtype=np.uint8)
checker[::2, ::2] = 255
checker[1::2, 1::2] = 255
```

```python
scale = 40.0
up_near  = cv.resize(checker, None, fx=scale, fy=scale, interpolation=cv.INTER_NEAREST)
up_lin   = cv.resize(checker, None, fx=scale, fy=scale, interpolation=cv.INTER_LINEAR)
up_cubic = cv.resize(checker, None, fx=scale, fy=scale, interpolation=cv.INTER_CUBIC)
```

```python
print(checker.shape)
print(up_near.shape)
print(up_lin.shape)
print(up_cubic.shape)
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.imshow(checker,cmap="gray",interpolation='none')
plt.subplot(2,2,2)
plt.imshow(up_near,cmap="gray",interpolation='none')
plt.subplot(2,2,3)
plt.imshow(up_lin,cmap="gray",interpolation='none')
plt.subplot(2,2,4)
plt.imshow(up_cubic,cmap="gray",interpolation='none')
```

```python
img_color = cv.imread("Cara2.jpg")
scale = 40.0
up_near  = cv.resize(img_color, None, fx=scale, fy=scale, interpolation=cv.INTER_NEAREST)
up_lin   = cv.resize(img_color, None, fx=scale, fy=scale, interpolation=cv.INTER_LINEAR)
up_cubic = cv.resize(img_color, None, fx=scale, fy=scale, interpolation=cv.INTER_CUBIC)
```

```python
print(img_color.shape)
print(up_near.shape)
print(up_lin.shape)
print(up_cubic.shape)
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.imshow(cv.cvtColor(img_color, cv.COLOR_BGR2RGB),interpolation='none')
plt.subplot(2,2,2)
plt.imshow(cv.cvtColor(up_near, cv.COLOR_BGR2RGB),interpolation='none')
plt.subplot(2,2,3)
plt.imshow(cv.cvtColor(up_lin, cv.COLOR_BGR2RGB),interpolation='none')
plt.subplot(2,2,4)
plt.imshow(cv.cvtColor(up_cubic, cv.COLOR_BGR2RGB),interpolation='none')
```

```python
# Seleccionar una región más pequeña, del ojo por ejemplo 
x, y, w, h = 2000, 5000, 1000, 1000  

crop_near = up_near[y:y+h, x:x+w]
crop_lin  = up_lin[y:y+h, x:x+w]
crop_cubic = up_cubic[y:y+h, x:x+w]

plt.figure(figsize=(12,4))
plt.subplot(1,3,1); plt.title("Vecino")
plt.imshow(cv.cvtColor(crop_near, cv.COLOR_BGR2RGB))
plt.subplot(1,3,2); plt.title("Bilineal")
plt.imshow(cv.cvtColor(crop_lin, cv.COLOR_BGR2RGB))
plt.subplot(1,3,3); plt.title("Bicubico")
plt.imshow(cv.cvtColor(crop_cubic, cv.COLOR_BGR2RGB))
plt.show()
```


## Ejercicio 3 — Composición afín: rotación + escala + traslación
Genera una matriz afín que combine las tres operaciones y aplícala con `warpAffine`.


```python
h, w = img_color.shape[:2]
center = (w/2, h/2)

M_rot = cv.getRotationMatrix2D(center=center, angle=30, scale=0.75)
M33 = np.vstack([M_rot, [0,0,1]])
T = np.array([[1,0,50],[0,1,-30],[0,0,1]], dtype=np.float32)
M = T @ M33
M_affine = M[:2,:]

out_affine = cv.warpAffine(img_color, M_affine, (w,h), flags=cv.INTER_LINEAR, borderMode=cv.BORDER_REFLECT101)

plt.figure(figsize=(8,4))
ax1 = plt.subplot(1,2,1); 
plt.imshow(cv.cvtColor(img_color, cv.COLOR_BGR2RGB))
plt.title('Original')
ax2 = plt.subplot(1,2,2); 
plt.imshow(cv.cvtColor(out_affine, cv.COLOR_BGR2RGB))
plt.title('Afin: R+S+T')
plt.tight_layout(); plt.show()
```

## Ejercicio 4 — Transformación proyectiva (homografía)
Define 4 puntos de origen (`src`) que formen un cuadrilátero y mapea a un rectángulo (`dst`). 
Usa `getPerspectiveTransform` + `warpPerspective`.

```python
img_color = cv.imread("chess.png")
```

```python

src = np.float32([[60,60],[280,50],[300,220],[40,200]])
W, H = 240, 320
dst = np.float32([[0,0],[W,0],[W,H],[0,H]])

Hmat = cv.getPerspectiveTransform(src, dst)
rectified = cv.warpPerspective(img_color, Hmat, (W,H), flags=cv.INTER_CUBIC, borderMode=cv.BORDER_REPLICATE)


plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(cv.cvtColor(img_color, cv.COLOR_BGR2RGB),interpolation='none')
plt.title("Original")
plt.subplot(1,2,2)
plt.imshow(cv.cvtColor(rectified, cv.COLOR_BGR2RGB),cmap="gray",interpolation='none')
plt.title("Rectificado (proyectiva)")
plt.tight_layout()
```


## Ejercicio 4 (Opcional) — Swirl
La transformación swirl está implementada en el paquete scikit-image


```python
from skimage.transform import swirl

image = img_bw
print(img_bw.shape)
swirled = swirl(image, rotation=0, strength=10, radius=500)

fig, (ax0, ax1) = plt.subplots(
    nrows=1, ncols=2, figsize=(8, 3), sharex=True, sharey=True
)

ax0.imshow(image, cmap=plt.cm.gray)
ax0.axis('off')
ax1.imshow(swirled, cmap=plt.cm.gray)
ax1.axis('off')

plt.show()
```
