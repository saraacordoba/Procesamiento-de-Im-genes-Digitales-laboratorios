```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Ejercicio 1: DETECCIÓN DE BORDES

```python
# 1. Cargar la imagen en escala de grises
img = cv2.imread('poker.jpg', cv2.IMREAD_GRAYSCALE)

# ==========================================
# a) MÉTODO BÁSICO (Suavizado + Sobel + Umbral)
# ==========================================
# El filtro de Sobel calcula la primera derivada de la imagen para encontrar el gradiente 
# de intensidad de los píxeles. Mide los cambios bruscos de brillo en las direcciones 
# horizontal (X) y vertical (Y). Al calcular el módulo del gradiente sqrt(SobelX^2 + SobelY^2), 
# obtenemos la magnitud de la variación de gris, destacando las zonas de posibles bordes.

# a. Suavizado gaussiano para reducir ruido antes de derivar
img_blur = cv2.GaussianBlur(img, (5, 5), 0)

# b. Operador de Sobel en X e Y
sobelx = cv2.Sobel(img_blur, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(img_blur, cv2.CV_64F, 0, 1, ksize=3)

# Módulo del gradiente: magnitud del cambio de intensidad
mod_grad = cv2.magnitude(sobelx, sobely)

# Normalizar el gradiente al rango [0, 255] para poder visualizarlo y umbralizarlo
mod_grad_uint8 = cv2.normalize(mod_grad, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

# c. Umbralización para binarizar los bordes
T_sobel = 50  # Umbral arbitrario de ejemplo
_, bordes_sobel = cv2.threshold(mod_grad_uint8, T_sobel, 255, cv2.THRESH_BINARY)

# ==========================================
# b) MÉTODO DE CANNY
# ==========================================
# Canny es un operador óptimo y multinivel que realiza 4 pasos clave:
# 1. Suavizado con filtro Gaussiano para eliminar el ruido.
# 2. Cálculo del gradiente y su dirección (similar a Sobel).
# 3. Supresión de no-máximos: adelgaza los bordes dejando solo las líneas de 1 píxel de ancho.
# 4. Umbralización por histeresis (con dos umbrales, low y high): los bordes por encima del 
#    umbral alto se aceptan inmediatamente, y los que están entre ambos umbrales solo se 
#    conservan si están conectados directamente con un borde seguro.

thresh_low = 50
thresh_high = 150
bordes_canny = cv2.Canny(img_blur, thresh_low, thresh_high)

# ==========================================
# VISUALIZACIÓN DE RESULTADOS
# ==========================================
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original (Gris)')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(mod_grad_uint8, cmap='gray')
plt.title('Módulo del Gradiente (Sobel)')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(bordes_sobel, cmap='gray')
plt.title(f'Método Básico: Sobel + Umbral (T={T_sobel})')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(bordes_canny, cmap='gray')
plt.title(f'Detector de Canny ({thresh_low}, {thresh_high})')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# EJERCICIO 2: UMBRALIZACIÓN GLOBAL

```python
# 1. Cargar la imagen original a color y en escala de grises
img_bgr = cv.imread('estrellas2.png')
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
img_gris = cv.cvtColor(img_bgr, cv.COLOR_BGR2GRAY)

# ==========================================
# a) Umbralizar la imagen
# ==========================================
# Usamos el método de Otsu para obtener el umbral óptimo de forma automática: Otsu es un método automático que decide 
# el umbral perfecto para separar las estrellas del fondo y dejar la imagen únicamente con 0s y 255s (blanco y negro puro).
# O un umbral simple (ej. T=127) según el contraste de las estrellas frente al fondo
ret, img_binaria = cv.threshold(img_gris, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

# Si las estrellas son oscuras sobre fondo claro, invertimos el umbral:
# ret, img_binaria = cv.threshold(img_gris, 0, 255, cv.THRESH_BINARY_INV + cv.THRESH_OTSU)

# ==========================================
# b) Localizar los contornos con cv.findContours
# ==========================================
# cv.RETR_EXTERNAL: Recupera únicamente los contornos externos
# cv.CHAIN_APPROX_SIMPLE: Comprime segmentos horizontales, verticales y diagonales
# cv.RETR_EXTERNAL (Modo de recuperación): Le dice a OpenCV que solo nos interesan los bordes exteriores de las estrellas. 
# Si una estrella tuviera un "agujero" en el centro o ruido interno, ignorará el contorno de dentro y solo se quedará con el 
# contorno de fuera.
# cv.CHAIN_APPROX_SIMPLE (Método de aproximación): Si no lo usáramos (CHAIN_APPROX_NONE), OpenCV guardaría todos los píxeles del borde
# de una línea recta. Con CHAIN_APPROX_SIMPLE, comprueba si la línea es recta y guarda únicamente las esquinas.

contornos, jerarquia = cv.findContours(img_binaria, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

print(f"Número total de contornos detectados: {len(contornos)}")

# ==========================================
# c) Calcular el área de cada contorno con cv.contourArea
# ==========================================
# Copiamos la imagen original para dibujar los resultados
img_resultado = img_rgb.copy()

# Filtramos los contornos de área mayor a 1000 píxeles
contornos_grandes = []
for cnt in contornos:
    area = cv.contourArea(cnt)
    if area > 1000:
        contornos_grandes.append(cnt)

print(f"Número de estrellas con área > 1000 píxeles: {len(contornos_grandes)}")

# ==========================================
# d) Visualizar los contornos seleccionados en rojo
# ==========================================
# cv.drawContours(imagen, contornos, -1 (todos), color RGB=(255, 0, 0), grosor=2)
cv.drawContours(img_resultado, contornos_grandes, -1, (255, 0, 0), 2)

# Mostrar la comparación de imágenes
plt.figure(figsize=(14, 6))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_binaria, cmap='gray')
plt.title('Imagen Umbralizada')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_resultado)
plt.title('Estrellas (Área > 1000 px) en Rojo')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# EJERCICIO 3. UMBRALIZACIÓN ADAPTATIVA

```python
# 1. Cargar la imagen en escala de grises
img = cv.imread('violin.jpg', cv.IMREAD_GRAYSCALE)

# Opcional: Filtro suave para reducir ruido antes de umbralizar
img_blur = cv.GaussianBlur(img, (5, 5), 0)

# ==========================================
# a) Umbralización de Otsu (Global)
# ==========================================
# Explicación de Otsu:
# Calcula automáticamente un ÚNICO umbral T para toda la imagen maximizando 
# la varianza entre las clases (fondo u objeto). Falla si hay iluminaciones desiguales.
val_otsu, img_otsu = cv.threshold(img_blur, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)
print(f"Umbral calculado por Otsu: {val_otsu}")

# ==========================================
# b) Umbralización Adaptativa (Local)
# ==========================================
# Explicación de la Umbralización Adaptativa:
# En lugar de usar un umbral global, calcula un umbral T(x,y) DIFERENTE para cada píxel 
# analizando una pequeña vecindad o ventana alrededor de él.

# Parámetros clave:
# - maxValue = 255 (valor asignado a píxeles que superan el umbral)
# - adaptiveMethod: 
#     * ADAPTIVE_THRESH_MEAN_C: Umbral = media aritmética de los vecinos - C
#     * ADAPTIVE_THRESH_GAUSSIAN_C: Umbral = suma ponderada gaussiana de vecinos - C
# - thresholdType = THRESH_BINARY
# - blockSize = 11 (tamaño de la ventana de vecindad, debe ser un número IMPAR)
# - C = 2 (constante que se resta a la media o suma gaussiana para afinar el resultado)

blockSize = 11
C = 2

# 1. Adaptativo basado en la Media (Mean)
img_adapt_mean = cv.adaptiveThreshold(
    img_blur, 255, cv.ADAPTIVE_THRESH_MEAN_C, cv.THRESH_BINARY, blockSize, C
)

# 2. Adaptativo basado en distribución Gaussiana (Gaussian)
img_adapt_gaussian = cv.adaptiveThreshold(
    img_blur, 255, cv.ADAPTIVE_THRESH_GAUSSIAN_C, cv.THRESH_BINARY, blockSize, C
)

# ==========================================
# c) Visualización y comparación de resultados
# ==========================================
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img_otsu, cmap='gray')
plt.title(f'Otsu Global (T={int(val_otsu)})')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img_adapt_mean, cmap='gray')
plt.title(f'Adaptativo Mean (block={blockSize}, C={C})')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img_adapt_gaussian, cmap='gray')
plt.title(f'Adaptativo Gaussian (block={blockSize}, C={C})')
plt.axis('off')

plt.tight_layout()
plt.show()
```

1. Otsu Global ($T=49$): La imagen presenta un claro degradado de iluminación (la luz proviene principalmente de la derecha). Al aplicar un único umbral global ($T=49$), el algoritmo es incapaz de compensar la sombra de la izquierda, considerando erróneamente que la mitad sombra del instrumento forma parte del fondo.

2. Adaptativo Mean ($\text{blockSize}=11$, $C=2$): Al calcular el umbral localmente en ventanas de $11 \times 11$ píxeles, se adapta perfectamente a las diferencias de luz entre la izquierda y la derecha. Sin embargo, al usar la media simple de los píxeles vecinos, detecta variaciones menores de textura en la madera y produce más ruido de fondo ("puntos" negros) en el interior del cuerpo.

3. Adaptativo Gaussian ($\text{blockSize}=11$, $C=2$): trazos más limpios y finos. La ponderación gaussiana asigna un peso mayor a los píxeles más cercanos al centro de la ventana local. Esto atenúa las pequeñas fluctuaciones locales, ofreciendo el resultado más equilibrado y con menor ruido para delimitar la geometría del violín.   

# EJERCICIO 4. K-MEANS

```python
# 1. Cargar la imagen a color
img_bgr = cv.imread('frutas_varias.jpg')
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

# 2. Reestructurar la imagen como una matriz 2D de píxeles (N_píxeles x 3 canales RGB)
# K-means necesita que cada píxel sea una muestra de 3 características (R, G, B)
pixels = img_rgb.reshape((-1, 3))
pixels = np.float32(pixels)

# 3. Criterios de parada para el algoritmo K-means:
# Detenerse si se alcanzan 100 iteraciones o si los centroides se mueven menos de epsilon = 0.2
criteria = (cv.TERM_CRITERIA_EPS + cv.TERM_CRITERIA_MAX_ITER, 100, 0.2)

# ==========================================
# Probar distintos valores de K (ej. K=2, K=3, K=4, K=5)
# ==========================================
valores_k = [2, 3, 4, 5]
resultados = []

for k in valores_k:
    # cv.kmeans(data, K, bestLabels, criteria, attempts, flags)
    compactness, labels, centers = cv.kmeans(
        pixels, k, None, criteria, 10, cv.KMEANS_RANDOM_CENTERS
    )
    
    # Convertir los centroides (colores medios de cada grupo) a valores enteros de 8 bits
    centers = np.uint8(centers)
    
    # Reemplazar la etiqueta de cada píxel por el color de su centroide
    segmented_data = centers[labels.flatten()]
    
    # Reestructurar la matriz de píxeles de vuelta a las dimensiones originales de la imagen
    segmented_image = segmented_data.reshape(img_rgb.shape)
    resultados.append((k, segmented_image))

# ==========================================
# VISUALIZACIÓN DE RESULTADOS
# ==========================================
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title('Imagen Original')
plt.axis('off')

for i, (k, seg_img) in enumerate(resultados):
    plt.subplot(2, 3, i + 2)
    plt.imshow(seg_img)
    plt.title(f'K-means (K = {k})')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

Con valores bajos ($K=2, K=3$), el algoritmo no logra segmentar las frutas porque se ve obligado a agrupar colores distintos (como el rojo de las fresas y el morado de los higos) en una sola categoría.   
El valor $K = 5$ resulta el más adecuado y óptimo para esta escena. Esto se debe a que la imagen no solo contiene 3 tipos de frutas, sino también las cajas verdes de plástico/madera y las sombras/fondo oscuro. Con $K=5$, cada fruta recibe su propio clúster cromático sin verse contaminada por el color de los recipientes o las sombras.

# EJERCICIO 5. WATERSHED

```python
# 1. Cargar la imagen a color y pasar a BGR/RGB/Gris
img_bgr = cv.imread('manzanas.png')
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)
img_gris = cv.cvtColor(img_bgr, cv.COLOR_BGR2GRAY)

# ==========================================
# Paso 1: Binarización e eliminación de ruido
# ==========================================
# Aplicamos binarización de Otsu
_, thresh = cv.threshold(img_gris, 0, 255, cv.THRESH_BINARY_INV + cv.THRESH_OTSU)

# Operación morfológica (Apertura) para eliminar pequeños puntos de ruido
kernel = np.ones((3, 3), np.uint8)
opening = cv.morphologyEx(thresh, cv.MORPH_OPEN, kernel, iterations=2)

# ==========================================
# Paso 2: Determinación de zonas conocidas (Fondo y Objeto)
# ==========================================
# Área del fondo seguro (Dilatación de la máscara binaria)
sure_bg = cv.dilate(opening, kernel, iterations=3)

# Transformada de distancia para encontrar los centros de los objetos
dist_transform = cv.distanceTransform(opening, cv.DIST_L2, 5)

# Área del objeto seguro (Umbralizamos la transformada de distancia al 50-70% de su máximo)
ret, sure_fg = cv.threshold(dist_transform, 0.5 * dist_transform.max(), 255, 0)
sure_fg = np.uint8(sure_fg)

# Región desconocida (Zona límite donde Watershed debe trazar la frontera)
unknown = cv.subtract(sure_bg, sure_fg)

# ==========================================
# Paso 3: Creación de marcadores para Watershed
# ==========================================
# Etiquetado de componentes conexas en el primer plano seguro
ret, markers = cv.connectedComponents(sure_fg)

# Sumamos 1 a todas las etiquetas para que el fondo seguro sea 1 (y no 0)
markers = markers + 1

# Marcamos la región desconocida como 0
markers[unknown == 255] = 0

# ==========================================
# Paso 4: Aplicar el algoritmo Watershed
# ==========================================
# OpenCV modifica la matriz de marcadores marcando las fronteras con -1
markers_result = cv.watershed(img_rgb, markers)

# Copia para dibujar las líneas divisoras en la imagen
img_watershed = img_rgb.copy()
img_watershed[markers_result == -1] = [255, 0, 0]  # Bordes dibujados en rojo

# ==========================================
# VISUALIZACIÓN DE RESULTADOS
# ==========================================
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title('1. Imagen Original')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(thresh, cmap='gray')
plt.title('2. Binarización Otsu')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(dist_transform, cmap='gray')
plt.title('3. Transf. Distancia')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(sure_fg, cmap='gray')
plt.title('4. Centroides / Objetos Seguros')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(markers_result, cmap='tab20')
plt.title('5. Máscara de Marcadores')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(img_watershed)
plt.title('6. Resultado Watershed (Rojo)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

1. Binarización Otsu:Al aplicar la umbralización global con Otsu, las manzanas quedan separadas del fondo blanco.   Problema observado: Debido a los brillos/reflejos blancos en la piel de las manzanas, aparecen "agujeros" internos de color negro dentro de la máscara binaria.   

2. Transf. Distancia:La transformada de distancia calcula la cercanía de cada píxel blanco respecto al borde negro más cercano.   Los picos más brillantes (puntos más alejados del borde) indican los centros geométricos o "núcleos" de cada manzana.

3. Centroides / Objetos Seguros:Al aplicar un umbral sobre la transformada de distancia, aislamos las regiones del primer plano seguro (sure foreground).   Se aprecian 5 bloques independientes, uno por cada manzana, sirviendo como semillas sin solapamiento.

4. Máscara de Marcadores:Se etiquetan las regiones conexas asignando un identificador/color único a cada marcador (las 5 manzanas y el fondo verde). La zona gris indeterminada entre marcadores es por donde se iniciará el proceso de inundación.

5. Resultado Watershed (Rojo):El algoritmo simula la inundación a partir de los marcadores hasta encontrar la cresta máxima de gradiente, trazando las líneas divisorias en rojo (-1). Separa individualmente cada manzana a lo largo del contorno exacto donde se estaban tocando.
