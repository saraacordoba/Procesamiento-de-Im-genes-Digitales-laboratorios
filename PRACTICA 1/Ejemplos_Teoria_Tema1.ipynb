### Cambiar la resolución de una imagen

En procesamiento digital de imágenes, cambiar la resolución se realiza mediante **redimensionado** (*resizing*).  
Este proceso consiste en modificar el número de píxeles de la imagen, ya sea para **reducir** su tamaño (downsampling) o para **aumentarlo** (upsampling).  

La calidad del resultado depende del método de **interpolación** usado:

- 🔽 **Reducir resolución (downsampling):** se recomienda `INTER_AREA`, que calcula promedios y evita aliasing.  
- 🔼 **Aumentar resolución (upsampling):** se usan `INTER_LINEAR` (rápido, bilineal) o `INTER_CUBIC` (más suave, mejor calidad).  

El redimensionado no solo afecta la apariencia visual, sino también el tamaño de memoria y el tiempo de procesamiento, por lo que es una operación muy habitual en preprocesamiento de imágenes.


```python
# Obtener tamaño original
h, w = im_gris.shape[:2]
print("Resolución original:", w, "x", h)

# Redimensionar a la mitad
img_half = cv.resize(im_gris, (w//4, h//4), interpolation=cv.INTER_AREA) #la interpolación INTER_AREA es buena para reducir, calcula el valor del píxel como el promedio de los píxeles vecinos

# Redimensionar al doble
img_double = cv.resize(im_gris, (w*4, h*4), interpolation=cv.INTER_CUBIC) #la interpolación INTER_CUBIC es buena para ampliar, usa un método de interpolación más sofisticado. cv.INTER_LINEAR (bilinear) o cv.INTER_CUBIC (bicubic) funcionan mejor porque suavizan y generan transiciones más naturales.

# Mostrar resultados
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(im_gris, cmap = 'gray')
plt.title("Original\n{}x{}".format(w,h))
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(img_half, cmap = 'gray')
plt.title("Reducida\n{}x{}".format(w//2,h//2))
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(img_double, cmap = 'gray')
plt.title("Ampliada\n{}x{}".format(w*2,h*2))
plt.axis("off")

plt.show()
```

### Modelos de color CMYK

El modelo de color **CMYK** (Cyan, Magenta, Yellow, Key/Black) se utiliza principalmente en el ámbito de la **impresión**.  
A diferencia de RGB, que es **aditivo** (los colores se forman sumando luz), CMYK es un modelo **sustractivo**: los colores se generan a partir de pigmentos que absorben (sustraen) ciertas longitudes de onda de la luz blanca.

- **C** (Cyan) absorbe el rojo.  
- **M** (Magenta) absorbe el verde.  
- **Y** (Yellow) absorbe el azul.  
- **K** (Black) se añade para mejorar el contraste, producir negros más profundos y reducir el consumo de tinta de los tres colores anteriores.

En teoría, la combinación de C, M y Y al 100% produciría el negro, pero en la práctica suele dar un tono marrón oscuro. Por ello, se introduce un canal adicional **K** que asegura la reproducción fiel de sombras y contornos.

Este modelo es fundamental en el flujo de trabajo de **impresión digital y offset**, ya que las impresoras trabajan depositando tintas (no luz). Así, mientras que RGB es el estándar en pantallas, **CMYK es el estándar en impresión**.


```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# Separar canales en orden RGB (recordar que OpenCV usa BGR por defecto)
R, G, B = im_color[:,:,2], im_color[:,:,1], im_color[:,:,0]

# Normalizar a [0,1]
R = R.astype(float) / 255.0
G = G.astype(float) / 255.0
B = B.astype(float) / 255.0

# Paso 1: RGB -> CMY
C = 1 - R
M = 1 - G
Y = 1 - B

# Paso 2: calcular K
K = np.minimum(np.minimum(C, M), Y)

# Paso 3: ajustar C, M, Y evitando división por cero
C_new = np.zeros_like(C)
M_new = np.zeros_like(M)
Y_new = np.zeros_like(Y)

mask = (K < 1)   # donde K != 1 (para evitar división por 0)
C_new[mask] = (C[mask] - K[mask]) / (1 - K[mask])
M_new[mask] = (M[mask] - K[mask]) / (1 - K[mask])
Y_new[mask] = (Y[mask] - K[mask]) / (1 - K[mask])
# donde K == 1, los tres canales quedan en 0 (ya están inicializados a cero)

# Reconstruir imagen CMYK
CMYK = np.stack((C_new, M_new, Y_new, K), axis=2)

# Escalar a [0,255] para visualización
CMYK_vis = (CMYK * 255).astype(np.uint8)

# Mostrar resultados
fig, axes = plt.subplots(1, 5, figsize=(15,4))

axes[0].imshow(cv.cvtColor(im_color, cv.COLOR_BGR2RGB))
axes[0].set_title("RGB original")
axes[0].axis("off")

axes[1].imshow(CMYK_vis[:,:,0], cmap="gray")
axes[1].set_title("Cyan")
axes[1].axis("off")

axes[2].imshow(CMYK_vis[:,:,1], cmap="gray")
axes[2].set_title("Magenta")
axes[2].axis("off")

axes[3].imshow(CMYK_vis[:,:,2], cmap="gray")
axes[3].set_title("Amarillo")
axes[3].axis("off")

axes[4].imshow(CMYK_vis[:,:,3], cmap="gray")
axes[4].set_title("Negro (K)")
axes[4].axis("off")

plt.tight_layout()
plt.show()
```
